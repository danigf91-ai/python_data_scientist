# SpaceX Data Wrangling
Completed IBM capstone data-cleaning and landing-label workflow.

```python
import pandas as pd
data=pd.read_csv('dataset_part_1.csv')
data['LaunchSite'].value_counts()
data['Orbit'].value_counts()
```

```python
bad_outcomes={'False Ocean','False ASDS','None None','False RTLS','None ASDS'}
landing_class=[0 if outcome in bad_outcomes else 1 for outcome in data['Outcome']]
data['Class']=landing_class
data.to_csv('dataset_part_2.csv',index=False)
```
