# SpaceX EDA with Data Visualization
Completed IBM capstone visualization and feature-engineering workflow.

```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df=pd.read_csv('dataset_part_2.csv')
sns.catplot(x='FlightNumber',y='LaunchSite',hue='Class',data=df,aspect=2)
sns.catplot(x='PayloadMass',y='LaunchSite',hue='Class',data=df,aspect=2)
```

```python
orbit_success=df.groupby('Orbit')['Class'].mean()
sns.barplot(x=orbit_success.index,y=orbit_success.values)
plt.xticks(rotation=45)
```

```python
features=df[['FlightNumber','PayloadMass','Orbit','LaunchSite','Flights','GridFins','Reused','Legs','LandingPad','Block','ReusedCount','Serial']]
features_one_hot=pd.get_dummies(features,columns=['Orbit','LaunchSite','LandingPad','Serial']).astype('float64')
features_one_hot.to_csv('dataset_part_3.csv',index=False)
```
